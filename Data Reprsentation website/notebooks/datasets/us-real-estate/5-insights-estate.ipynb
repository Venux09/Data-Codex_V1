```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import sympy as sp 
```

```python
Estate_Data = pd.read_csv('cleaned_Estate.csv')
```

```python
# Keep only rows where the price is greater than a reasonable minimum
Estate_Data = Estate_Data[Estate_Data['price'] > 1000]
```

```python
#The most expensive area to buy the houses 
top_10_cities = Estate_Data.groupby('city')['price'].median().sort_values(ascending = False).head(10)
```

```python
#the top 10 cities with lowest median houses prices 
Low_10_cities = Estate_Data.groupby('city')['price'].median().sort_values(ascending = True).head(10)
```

```python
#so now we can also find the which states have most active sale market like which state have more more percentage to the ratio of there house sale 
#so here we are calculating percentage because of number of the houses cities are different and other measure 
```

```python
market_percentage = Estate_Data.groupby('state')['status'].value_counts(normalize = True)*100
market_percentage
```

```python
active_market = market_percentage.loc[:,'for_sale'].sort_values(ascending=False)
```

```python
print(f'The Top State with highest percentage of the active market it :\n{active_market.head(10)}')
```

```python
"so here comes the problem that data in the data maybe there is hoow with 1 house for sale and nothing sold then we have to think about the bias to think about we have to set limit atleast every entry should have atleast 500 of house for sale and sold as totle"
```

```python
#solution 
state_summary = Estate_Data.groupby('state')['status'].agg(
    total_listing = 'count',
    percentage_for_sale  = lambda s: (s=='for_sale').mean()*100
    )#calculate the count of the data and the percentage on the data we have done the multiple math so we choose aggregation here 
reliable_market = state_summary[state_summary['total_listing']>=100]
active_reliable_market = reliable_market.sort_values(by='percentage_for_sale',ascending = False).head(10)
```

```python
print(f'The Top State with highest percentage of the active market it :\n{active_reliable_market.head(10)}')
```
