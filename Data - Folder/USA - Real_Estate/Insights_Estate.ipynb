# Insights of the Cleaned Estate dataset 

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import sympy as sp 
```

```python
Estate_Data = pd.read_csv('cleaned_Estate.csv')
```

```python
# Keep only rows where the price is greater than a reasonable minimum
Estate_Data = Estate_Data[Estate_Data['price'] > 1000]
```

```python
"We basically did this because when we finded the answer of the problem with Top 10 cheapest cities with cheapest median houses we found out that there were houses with 0 dollar vlaue so we did before finding the answer of the quetion"
```

```python
#The most expensive area to buy the houses 
top_10_cities = Estate_Data.groupby('city')['price'].median().sort_values(ascending = False).head(10)
```

```python
print(f'THE TOP 10 cities with most expensive houses\n{top_10_cities}')
```

```python
"we are using the  median here because we want to find the median value so that if we use the mean  then some property like 100m$ can't inflate the data "
```

```python
#the top 10 cities with lowest median houses prices 
Low_10_cities = Estate_Data.groupby('city')['price'].median().sort_values(ascending = True).head(10)
```

```python
print(f"The Top 10 cities with cheapest median houses\n{Low_10_cities})
```
