```python
#filling the not registered value of the bed and bath accordingn to the meadian of the region using groupby 
Estate_data['bed'] = Estate_data.groupby('zip_code')['bed'].transform(lambda x : x.fillna(x.median()) )
```

```python
Estate_data['bath'] = Estate_data.groupby('zip_code')['bath'].transform(lambda x : x.fillna(x.median()))
```

```python
#filling the acre_lot with the same logic 
Estate_data['acre_lot'] = Estate_data.groupby('zip_code')['acre_lot'].transform(lambda x : x.fillna(x.median()))
```

```python
Estate_data.isnull().sum()
```

```python
#filling the missing values with the meadian of that particular zip cod e
Estate_data['house_size'] = Estate_data.groupby('zip_code')['house_size'].transform(lambda x: x.fillna(x.median()))
```

```python
"So here i am creating a has been sold column it is very common that house are either new so we are saying that if sold 1 if not sold 0 in order to predict that house has more value since it is new "
```
